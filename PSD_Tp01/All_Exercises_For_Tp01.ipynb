```python
def celsius_to_fahrenheit (c, round_to = 1):
    value = c * 9/5 + 32
    return round (value, round_to)

def fahrenheit_to_celsuis (f, round_to = 1):
    value = (f - 32) * 5/9
    return round (value, round_to)

# (a)
print ("\n(a) -- Celsius to Fahrenheit\n")
for c in range (0, 101, 20):
    f = celsius_to_fahrenheit(c)

    print (f"{c}\t==>\t{f}")

# (b)
print ("\n(b) -- Convert 98.6F to C\n")
temp_F = 98.6
temp_C = fahrenheit_to_celsuis(temp_F)

print (f"Convert {temp_F} F to Celsisus equal {temp_C}")

# (c)
print ("\n(c) -- Turn 37C to F and back again\n")
f = celsius_to_fahrenheit(37)
c = fahrenheit_to_celsuis(f)

print (f"37C to F equal {f} and {f} to C equal {c}")

# (d)
print ("\n(d) -- Call one function with round to 3\n")
r = celsius_to_fahrenheit(23.42, round_to = 3)

print (f"{r}\n")
```

```python
# PART B
def safe_int(text):
    try:
        return int(float(text.strip()))
    except (ValueError, OverflowError, AttributeError):
        return None


for sample in ["42", "-8", "3.9", "abc", ""]:
    print(f"{sample} -> {safe_int(sample)}")

# PART C
first = safe_int(input("\nEnter the first number: "))
second = safe_int(input("\nEnter the second number: "))

if first is None or second is None:
    print("\nSorry, one of the values is not a number. Please try again.")
else:
    print("\nThe sum is:", first + second)
```

```python
# (a)
def is_leap(year):
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

a = [200, 400]
for i in a:
    leap = is_leap(i)
    print (leap)

# (b)
def days_in_month (month, year):
    if month < 1 or month > 12:
        return None
    if month == 2:
        return 29 if is_leap(year) else 28
    if month in (4, 6, 9 ,11):
        return 30
    return 31

print (f"Feb 2024: {days_in_month(2, 2024)} | "
       f"Feb 2026: {days_in_month(2, 2026)} | "
       f" Apr: {days_in_month(4, 2026)} | "
       f"Month 13: {days_in_month(13, 2026)}")

# (c) check leap year
for y in (1900, 2000, 2024, 2026):
    print (y, "Leap" if is_leap(y) else "Not leap")

# (d) count leap year
count = 0
for y in range (2000, 2101):
    if is_leap(y):
        count += 1

print(f"Leap years from 2000 to 2100: {count}")
```

```python
def digit_sum (n):
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total

def revers_number (n):
    revers = 0
    while n > 0:
        revers = revers * 10 + n % 10
        n //= 10
    return revers

def is_palindrome (n):
    return n == revers_number(n)

# (a) Test the three functions with 2026, 12321 and 9875

for t in (2026, 12321, 9875):
    print (f"{t}: Digit sum = {digit_sum(t)}, Reversed = {revers_number(t)}, Palindrome = {is_palindrome(t)}")

# (b) Count how many palindromes there are from 10 to 999.

count = 0
for i in range (10, 1000):
    if is_palindrome(i):
        count += 1

print (f"Palindromes from 10 to 999: {count}")
```

```python
# (a) Descibe data
def describe (*numbers, decimals = 2, label = 'Data'):
    if not numbers:
        print (f"{label}: not number")
        return None
    count = len(numbers)
    minimum = min(numbers)
    maximum = max(numbers)
    mean = sum(numbers) / count

    print (f"{label}: n = {count} "
           f"min = {minimum} "
           f"max = {maximum} "
           f"mean = {mean:.{decimals}f} ")

    return count, minimum, maximum, round(mean, decimals)

describe(12, 7, 30, 4)
describe(3.14159, 2.71828, decimals=3, label="Constants")
describe()

# (b) 
def make_tag(text, **style):
    attributes = " ".join(
        f'{key} = "{value}"' for key, value in style.items()
    )
    return f"<span {attributes}>{text}</span>"

print(make_tag("Hello", color="red", size=14))

# (c) Write two lamda
square = lambda x: x ** 2
is_adult = lambda age: age >= 18
print(square(9), is_adult(17), is_adult(21))
```

```python
# (a) Prime number function
def is_prime(n):
    if n <= 1:
        return False
        
    primes = (2, 3, 5, 7, 11, 13, 17, 19)
    if n in primes:
        return True
        
    for i in primes:
        if n % i == 0:
            return False
            
    return True

# (b) Print all prime below 50, and count prime below 100
print("Primes Below 50 : ")
for n in range(1, 50): 
    if is_prime(n):
        print(n, end=" ") 
print ()

# Count prime below 100
count = 0
for i in range(1, 101): 
    if is_prime(i):
        count += 1

print (f"Count Prime number between 1 - 100: {count}")

#Goldbach function
def goldbach(n): 
    for p in range(2, n): 
        q = n - p 
        if is_prime(p) and is_prime(q): 
            return p, q
    return None 

# (c) Pairs for 4, 6, ..., 30
for n in range(4, 31, 2):
    p, q = goldbach(n)
    print(f"{n} = {p} + {q}")

# (d) Check every even number from 4 to 10000, stopping at the first failure
holds = True
for n in range(4, 10001, 2):
    if goldbach(n) is None:
        holds = False
        break
print(f"Goldbach holds for every even number from 4 to 10000: {holds}")
```
